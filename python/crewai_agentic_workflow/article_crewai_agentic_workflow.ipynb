## 1. Get the API Keys

```python
# Import the packages
import os
from pathlib import Path
from dotenv import load_dotenv

# Specify the root folder
ROOT_DIR = Path.cwd().parents[2]

# Get the .env path
ENV_PATH = ROOT_DIR / ".env"

# Load the .env
load_dotenv(dotenv_path=ENV_PATH)

# Get the API keys
GEMINI_API_KEY = os.getenv("GEMINI_API_KEY")
SERPER_API_KEY = os.getenv("SERPER_API_KEY")
```

## 2. Create the Research Agent & Task

```python
# Import the package
from crewai import LLM

# Configure the LLM
llm = LLM(
    model="gemini/gemini-2.5-flash",
    api_key=GEMINI_API_KEY
)
```

```python
# Import the package
from crewai_tools import SerperDevTool

# Initialize the SerperDevTool with the API key
search_tool = SerperDevTool(api_key=SERPER_API_KEY)
```

```python
# Import the package
from crewai import Agent

# Create a research agent
researcher = Agent(
    role="Web Researcher",
    goal="Find accurate and relevant information on the research topic",
    backstory="""
    You are an experienced research assistant.
    You search the web, evaluate sources, and extract useful
    information relevant to the research question.
    """,
    llm=llm,
    tools=[search_tool],
    verbose=True
)
```

```python
# Import the package
from crewai import Task

# Create a research task
research_task = Task(
    description="""
    Research the following question:

    {question}

    Find relevant information from reliable web sources.
    Identify concrete applications and provide supporting evidence.
    """,
    expected_output="""
    A structured summary of the main applications of RAG in healthcare,
    including supporting sources and key findings.
    """,
    agent=researcher
)
```

## 3. Create the Analysing Agent & Task

```python
# Create an analyst agent
analyst = Agent(
    role="Research Analyst",
    goal="Analyse research findings and identify important patterns and conclusions",
    backstory="""
    You are an analytical researcher.
    You examine research findings, compare evidence,
    identify recurring themes, and distinguish strong
    evidence from weak evidence.
    """,
    llm=llm,
    verbose=True
)
```

```python
# Create an analysis task
analysis_task = Task(
    description="""
    Analyse the research produced by the research agent.

    Identify:
    - The main applications of RAG in healthcare
    - Recurring themes
    - Important differences between applications
    - Strength of supporting evidence

    Do not introduce claims that are not supported by the research.
    """,
    expected_output="""
    A structured analysis of the research findings, organised by
    major themes and applications.
    """,
    agent=analyst
)
```

## 4. Create the Writing Agent & Task

```python
# Create a writer agent
writer = Agent(
    role="Research Writer",
    goal="Turn research findings into a clear and well-structured report",
    backstory="""
    You are an experienced research writer.
    You turn complex research into clear, concise,
    well-organised reports while preserving factual accuracy.
    """,
    llm=llm,
    verbose=True
)
```

```python
# Create a writing task
writing_task = Task(
    description="""
    Write a concise research report based on the research
    and analysis produced by the previous agents.

    The report should include:

    1. Introduction
    2. Main applications
    3. Key findings
    4. Important limitations
    5. Sources

    Clearly distinguish evidence from interpretation.
    """,
    expected_output="""
    A polished research report with clear sections and source references.
    """,
    agent=writer
)
```

## 5. Run the Crew

```python
# Import the package
from crewai import Crew

# Create a crew
crew = Crew(
    agents=[
        researcher,
        analyst,
        writer
    ],
    tasks=[
        research_task,
        analysis_task,
        writing_task
    ],
    verbose=True
)
```

```python
# Run the crew
result = await crew.kickoff_async(
    inputs={
        "question": "What are the applications of RAG in healthcare?"
    }
)

# Print the results
print(result)
```
