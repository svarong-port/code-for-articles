## 3 Create a Scraper

```python
# Import the package
from google.colab import userdata

# Get the API key
GEMINI_API_KEY = userdata.get("GEMINI_KEY_FOR_ARTICLE")
```

```python
# Import the packages
from langchain_google_genai import ChatGoogleGenerativeAI

# Set the LLM
llm = ChatGoogleGenerativeAI(
    api_key=GEMINI_API_KEY,
    model="gemini-2.5-flash",
    temperature=0.0,
    timeout=120,
)
```

```python
# Import the package
from parsera import Parsera

# Create a Parsera object
scraper = Parsera(
    model=llm,
    typed=True
)
```

## 4 Set Input for the Scraper

```python
# Set the web page to scrape
url = "https://books.toscrape.com/catalogue/category/books/science-fiction_16/index.html"
```

```python
# Set the elements to scrape
elements = {
    "title": {
        "description": "The title of the book",
        "type": "string"
    },
    "price_gbp": {
        "description": "The price of the book, in GBP",
        "type": "number"
    },
    "is_in_stock": {
        "description": "Whether the book is in stock",
        "type": "bool"
    },
    "full_image_url": {
        "description": "The abosolute URL path of the book's image",
        "type": "string"
    }
}
```

## 5 Run the Scraper

```python
# Run the scraper
result = scraper.run(
    url=url,
    elements=elements
)
```
