## 1. Connect to the Database

```python
# Import the package
from sqlalchemy import create_engine

# Connect to the database
engine = create_engine("sqlite:///chinook.sqlite")
```

## 2. List the Tables

```python
# Import the package
from sqlalchemy import inspect

# Get the inspector
inspector = inspect(engine)
```

```python
# List the table names
table_names = inspector.get_table_names()

# Print the table names
print(table_names)
```

```python
# Get column names
columns = inspector.get_columns("Invoice")

# Print the result
for col in columns:
    print(col["name"], col["type"])
```

## 3. Write a Query

```python
# Import the package
from sqlalchemy import text

# Set a query
query = text(
    """
    SELECT 
        InvoiceId,
        InvoiceDate,
        BillingCountry,
        Total
    FROM Invoice;  
    """
)
```

## 4. Load the Data

```python
# Import the package
import pandas as pd

# Load the table
df = pd.read_sql(
    query,
    engine
)
```

```python
# Inspect the df
df.head()
```

## 5. Analyse

```python
# Find mean total by country
(
    df
    .groupby("BillingCountry", as_index=False)
    .agg(MeanTotal=("Total", "mean"))
    .sort_values("MeanTotal", ascending=False)
)
```
