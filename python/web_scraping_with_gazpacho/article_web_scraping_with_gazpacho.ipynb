## 1. Get the Web Page

```python
# Load the package
import gazpacho as gz
```

```python
# Set the URL
url = "https://books.toscrape.com/catalogue/sapiens-a-brief-history-of-humankind_996/index.html"

# Load the web page
page = gz.get(url)
```

```python
# Inspect the page
print(page[:1000])
```

## 2. Create a Soup Object

```python
# Create a Soup object
soup = gz.Soup(page)
```

## 3. Extract the Content

```python
# Extract the book title
book_title = soup.find("h1").text

# Print the book title
print(book_title)
```

```python
# Extract the book price
book_price = soup.find("p", {"class": "price_color"})[0].text

# Print the book price
print(book_price)
```

```python
# Extract the book image path
book_image_path = soup.find("img")[0].attrs["src"]

# Append to the base URL
book_image_url = "https://books.toscrape.com/" + book_image_path.replace("../../", "")

# Print the image URL
print(book_image_url)
```
